```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load Titanic Dataset
titanic = sns.load_dataset("titanic")
```

```python
# 1. Countplot: Passenger Count by Embarked Port
plt.figure(figsize=(8,5))
sns.countplot(data=titanic, x="embarked", hue="sex")

plt.title("Passenger Count by Embarked Port")
plt.xlabel("Embarked Port")
plt.ylabel("Number of Passengers")
plt.legend(title="Sex")
plt.show()

print("Observation:")
print("- Southampton (S) has the highest number of passengers.")
print("- Male passengers are more than female passengers at all embarkation ports.\n")

```

```python
# 2. Swarmplot: Age Distribution across Classes
plt.figure(figsize=(10,6))
sns.swarmplot(data=titanic, x="pclass", y="age", hue="sex")

plt.title("Age Distribution across Passenger Classes")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.legend(title="Sex")
plt.show()

print("Observation:")
print("- First-class passengers include many adults and elderly people.")
print("- Third-class passengers have more younger passengers.")
print("- Male and female passengers are present in all classes.\n")
```

```python
# 3. Pie Chart: Gender Distribution

gender_counts = titanic["sex"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct="%1.1f%%",
    explode=(0.05, 0),
    startangle=90
)

plt.title("Gender Distribution of Passengers")
plt.show()

print("Observation:")
print("- Male passengers form the majority of the dataset.")
print("- The exploded slice highlights the male passenger group.\n")
```

```python
# 4. Heatmap: Correlation Matrix

corr = titanic[["age", "fare", "pclass", "sibsp", "parch", "survived"]].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Matrix")
plt.show()

print("Observation:")
print("- Fare and passenger class have a strong negative correlation.")
print("- Survival has a positive relationship with fare.")
print("- Passenger class has a negative relationship with survival.\n")

```

```python
# 5. Violin Plot: Fare by Class and Gender
plt.figure(figsize=(10,6))
sns.violinplot(data=titanic, x="pclass", y="fare", hue="sex", split=True)

plt.title("Fare Distribution by Passenger Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.legend(title="Sex")
plt.show()

print("Observation:")
print("- First-class passengers generally paid higher fares.")
print("- Third-class passengers paid the lowest fares.")
print("- Fare distributions differ slightly between males and females.\n")
```

```python
# 6. Scatter Plot: Age vs Fare

plt.figure(figsize=(10,6))
sns.scatterplot(data=titanic, x="age", y="fare", hue="survived")

plt.title("Age vs Fare by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived")
plt.show()

print("Observation:")
print("- Passengers who paid higher fares had a greater chance of survival.")
print("- Most passengers with low fares are concentrated in the lower fare range.")
print("- Age alone does not show a strong relationship with fare.")
```
